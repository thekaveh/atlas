# Hello Ray

This notebook demonstrates direct Ray Client calls to the in-stack Ray cluster.

## Prerequisites

Ray Client requires the same Python minor version and Ray version in the notebook kernel and cluster. Atlas locks Ray to `2.56.0`; the Ray image uses Python `3.10`. The default JupyterHub kernel uses Python `3.13`, so these direct-client cells fail from that kernel.

Use a compatible Python `3.10` kernel with `ray[client]==2.56.0` to run the examples below. For the default kernel, follow the [Ray Jobs REST API example](https://github.com/thekaveh/atlas/blob/main/services/ray/README.md#6-troubleshooting) instead.

`RAY_ADDRESS` is set automatically when Ray is enabled; `ray.init()` reads it. If Ray is disabled, the first code cell reports that the address is missing.

```python
import os
import ray

ray_address = os.environ.get("RAY_ADDRESS", "").strip()
if not ray_address:
    raise RuntimeError(
        "RAY_ADDRESS is not set — Ray is disabled in this deployment. "
        "Re-run ./start.sh --ray-source ray-container-cpu (or ray-container-gpu) to enable it."
    )

ray.init(address=ray_address)
print("cluster resources:", ray.cluster_resources())
```

```python
@ray.remote
def double(x):
    return x * 2

results = ray.get([double.remote(i) for i in range(10)])
print(results)
```
